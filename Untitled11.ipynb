```python
import warnings
warnings.simplefilter(action='ignore', category=FutureWarning)
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder, MinMaxScaler

file_path = r"C:\Users\carnival\Downloads\archive(1)\StudentsPerformance.csv"
df = pd.read_csv(file_path)
df_cleaned = df.copy()

print("Shape:", df.shape)
print(df.head())
print(df.info())
print(df.describe())
```

```python
categorical_cols = [
    'gender',
    'race/ethnicity',
    'parental level of education',
    'lunch',
    'test preparation course'
]

le = LabelEncoder()
for col in categorical_cols:
    df_cleaned[col] = le.fit_transform(df_cleaned[col])
```

```python
score_cols = ['math score', 'reading score', 'writing score']

scaler = MinMaxScaler()
df_cleaned[score_cols] = scaler.fit_transform(df_cleaned[score_cols])
```

```python
df_cleaned['average_score'] = df[['math score', 'reading score', 'writing score']].mean(axis=1)

def classify(score):
    if score >= 80:
        return 'High'
    elif score >= 60:
        return 'Medium'
    else:
        return 'Low'

df_cleaned['performance_level'] = df_cleaned['average_score'].apply(classify)
```

```python
for col in score_cols:
    print(f"\n--- {col.upper()} ---")
    print("Mean:", df_cleaned[col].mean())
    print("Median:", df_cleaned[col].median())
    print("Std Dev:", df_cleaned[col].std())
    print("Min:", df_cleaned[col].min())
    print("Max:", df_cleaned[col].max())

    plt.figure(figsize=(6, 4))
    sns.boxplot(y=df_cleaned[col])
    plt.title(f'Boxplot of {col}')
    plt.show()

    plt.figure(figsize=(6, 4))
    sns.histplot(df_cleaned[col], kde=True, bins=20)
    plt.title(f'Histogram of {col}')
    plt.show()
```

```python
df_cleaned.to_csv("students_cleaned.csv", index=False)
print("✔️ Cleaned data saved as students_cleaned.csv")
```
